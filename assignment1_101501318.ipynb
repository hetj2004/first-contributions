```python
"""
Author: <Het Jasani>
Assignment: #1
"""

# String (str)
gym_member = "Sahil Nathani"  

# Float (float)
preferred_weight_kg = 20.5  

# Integer (int)
highest_reps = 25  

# Boolean (bool)
membership_active = True  

# Dictionary (dict) with string keys and tuple values containing three integers
workout_stats = {
    "Sahil": (30, 45, 20),   # Yoga: 30 min, Running: 45 min, Weightlifting: 20 min
    "Meet": (25, 40, 35),  # Yoga: 25 min, Running: 40 min, Weightlifting: 35 min
    "Kiltan": (20, 50, 30)  # Yoga: 20 min, Running: 50 min, Weightlifting: 30 min
}

# Loop through the dictionary to calculate total workout minutes for each friend
for friend, minutes in list(workout_stats.items()):  
    total_minutes = sum(minutes)  # Calculate the total workout minutes
    workout_stats[f"{friend}_Total"] = total_minutes  # Add new key-value pair

# List of lists (2D list) containing workout minutes for each friend
workout_list = [minutes for friend, minutes in workout_stats.items() if not friend.endswith("_Total")]

# Extract and print the minutes for yoga and running for all friends (first two columns)
yoga_running_minutes = [row[:2] for row in workout_list]
print("Yoga and Running Minutes for All Friends:", yoga_running_minutes)

# Extract and print the minutes for weightlifting for the last two friends (third column, last two rows)
weightlifting_last_two = [row[2] for row in workout_list[-2:]]
print("Weightlifting Minutes for Last Two Friends:", weightlifting_last_two)

# Loop through the dictionary to check for friends with total workout minutes >= 120
for friend, total_minutes in workout_stats.items():
    if friend.endswith("_Total") and total_minutes >= 120:
        original_name = friend.replace("_Total", "")
        print(f"Great job staying active, {original_name}!")

# Get user input for a friend's name
friend_name = input("Enter a friend's name: ")

# Check if the friend's name exists in the dictionary
if friend_name in workout_stats:
    workout_minutes = workout_stats[friend_name]
    total_minutes = workout_stats.get(f"{friend_name}_Total", "N/A")  # Get total if available
    print(f"{friend_name}'s Workout Minutes:")
    print(f"Yoga: {workout_minutes[0]} min, Running: {workout_minutes[1]} min, Weightlifting: {workout_minutes[2]} min")
    print(f"Total Workout Minutes: {total_minutes} min")
else:
    print(f"Friend {friend_name} not found in the records.")

# Extract total workout minutes from the dictionary
total_minutes_dict = {friend.replace("_Total", ""): minutes for friend, minutes in workout_stats.items() if friend.endswith("_Total")}

# Check if there are any records
if total_minutes_dict:
    # Find the friend with the highest and lowest total workout minutes
    max_friend = max(total_minutes_dict, key=total_minutes_dict.get)
    min_friend = min(total_minutes_dict, key=total_minutes_dict.get)

    # Print results
    print(f"The friend with the highest total workout minutes is {max_friend} with {total_minutes_dict[max_friend]} min.")
    print(f"The friend with the lowest total workout minutes is {min_friend} with {total_minutes_dict[min_friend]} min.")
else:
    print("No total workout data available.")
```
